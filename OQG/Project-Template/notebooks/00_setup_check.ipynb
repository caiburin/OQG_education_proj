## 3. Download your pair
Change the two tickers to your pair, then run the cell. It saves one CSV per asset in the `data/` folder.
**Never edit these files by hand.**

Keep the dates fixed (not "today") so your results don't change every time you re-run.

````python
TICKERS = ["SPY", "RSP"]    # <- change to your pair (exact Yahoo Finance tickers)
START = "2016-01-01"         # ten years of daily data
END = "2026-01-01"

for ticker in TICKERS:
    px = yf.download(ticker, start=START, end=END, auto_adjust=False, multi_level_index=False)
    px.to_csv(f"../data/{ticker}.csv")
    print(ticker, "saved:", len(px), "days")
````

## 4. Download the Fama-French factors (used in Week 7)

````python
import pandas_datareader.data as web

ff = web.DataReader("F-F_Research_Data_Factors_daily", "famafrench", start=START, end=END)[0]
mom = web.DataReader("F-F_Momentum_Factor_daily", "famafrench", start=START, end=END)[0]
mom.columns = ["Mom"]

factors = ff.join(mom) / 100     # percent -> decimal
factors.to_csv("../data/ff_factors.csv")
factors.tail()
````
